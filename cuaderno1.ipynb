```python
import pandas as pd
import random
```

```python
from faker import Faker
fake = Faker()

provincias = ['Loja', 'Pichincha', 'Guayas', 'Manabí', 'Azuay' ]
```

```python
lista = [{
    'Nombre': fake.name(),
    'Apellido': fake.last_name(),
    'Email': fake.email(),
    'Direccion': fake.address()

} for x in range(10000)]
```

```python
data = pd.DataFrame(lista)
print(data)
data['Edad'] = [fake.random.randint(30,40) for x in range(len(data))]
data['Provincias'] = [fake.random_element(provincias) for x in range(len(data))] 
data= data [['Nombre','Apellido','Email','Direccion','Edad','Provincias']]
print(data)
```

```python
data.to_csv("midata.csv", index=False)
```
