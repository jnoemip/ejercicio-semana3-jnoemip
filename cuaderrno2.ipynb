```python
import pandas as pd
import pygwalker as pyg
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('midata.csv')
walker=pyg.walk(df, kernel_computation=True)
print("visualizar pygwalker")
```

```python
df['Edad'].value_counts().sort_index().plot(kind='bar', color='skyblue')
plt.title('Regristro por Edad')
plt.xlabel('Edad')
plt.xlabel('Numero de Registros')
```

```python

df['Provincias'].value_counts().sort_index().plot(kind='bar', color='green')
plt.title('Regristro por Provincia')
plt.xlabel('Provincias')
plt.xlabel('Numero de Registros')
```
